```python
# =============================================================================
# CONVORELENET — MODULE 5 UPDATED MASTER PIPELINE
# =============================================================================
#
# DATA:
#   X         : (N, 22, 640)
#   y         : (N,)
#   subjects  : (N,)
#
# CURRENT DATASET:
#   BCI-IV-2a + EEGMMIDB
#   22 EEG channels
#   640 samples
#   160 Hz
#   3 classes
#
# PIPELINE:
#
#                  RAW EEG
#                    |
#             Trial normalization
#                    |
#          +---------+---------+
#          |                   |
#    Temporal branch      Spatial branch
#          |                   |
#          +---------+---------+
#                    |
#             Feature fusion
#                    |
#             Linear -> 128
#                    |
#          Positional Encoding
#                    |
#        8-Layer Transformer
#          4 attention heads
#                    |
#          3-Layer BiLSTM
#                    |
#                MLP
#                    |
#               3 classes
#
# LOSO:
#   source subjects = all except target
#   target calibration = target training subset
#   target test       = completely held-out subset
#
# =============================================================================

import os
import gc
import copy
import time
import math
import random
import warnings

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import (
    TensorDataset,
    DataLoader,
    WeightedRandomSampler
)

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    cohen_kappa_score,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# =============================================================================
# 0. CONFIGURATION
# =============================================================================

SEED = 42

# -----------------------------
# Dataset
# -----------------------------
EXPECTED_CHANNELS = 22
EXPECTED_SAMPLES = 640
EXPECTED_CLASSES = 3
EXPECTED_FS = 160.0

# -----------------------------
# Experiment
# -----------------------------
RUN_FULL_LOSO = False

# For initial debugging:
# ["BCI-IV-2a::S01"]
TARGET_SUBJECTS = [
    "BCI-IV-2a::S01"
]

# Set to None to automatically select first unique target
# TARGET_SUBJECTS = None

# -----------------------------
# Source pretraining
# -----------------------------
PRETRAIN_EPOCHS = 150
PRETRAIN_BATCH_SIZE = 64
PRETRAIN_LR = 1e-3
PRETRAIN_WEIGHT_DECAY = 1e-4

# -----------------------------
# Target adaptation
# -----------------------------
TARGET_CALIBRATION_RATIO = 0.20

TARGET_FT_EPOCHS = 75
TARGET_FT_BATCH_SIZE = 8
TARGET_FT_LR = 5e-5
TARGET_FT_WEIGHT_DECAY = 1e-4

# -----------------------------
# Data normalization
# -----------------------------
USE_TRIAL_NORMALIZATION = True

# Clip extreme standardized samples
NORMALIZE_CLIP = 8.0

# -----------------------------
# Optimization
# -----------------------------
GRAD_CLIP = 1.0

# Prevent exploding Transformer activations
USE_GRAD_CHECKPOINTING = False

# -----------------------------
# Architecture
# -----------------------------
D_MODEL = 128
N_HEADS = 4
N_TRANSFORMER_LAYERS = 8

LSTM_HIDDEN = 128
N_LSTM_LAYERS = 3

CONV_DROPOUT = 0.40
LSTM_DROPOUT = 0.30
MLP_DROPOUT = 0.50

NUM_CLASSES = EXPECTED_CLASSES

# -----------------------------
# Diagnostics
# -----------------------------
RUN_MICRO_OVERFIT_TEST = True
MICRO_OVERFIT_N_PER_CLASS = 8
MICRO_OVERFIT_EPOCHS = 100

# -----------------------------
# Reproducibility
# -----------------------------
os.environ["PYTHONHASHSEED"] = str(SEED)

random.seed(SEED)
np.random.seed(SEED)

torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("\n" + "=" * 80)
print("CONVORELENET — UPDATED MODULE 5 MASTER PIPELINE")
print("=" * 80)

# =============================================================================
# 1. DEVICE
# =============================================================================

if torch.backends.mps.is_available():
    DEVICE = torch.device("mps")
elif torch.cuda.is_available():
    DEVICE = torch.device("cuda")
else:
    DEVICE = torch.device("cpu")

print("Device:", DEVICE)

# =============================================================================
# 2. VALIDATE EXISTING DATASET VARIABLES
# =============================================================================

required_names = ["X", "y"]

missing = [
    name for name in required_names
    if name not in globals()
]

if len(missing) > 0:
    raise RuntimeError(
        "Missing required dataset variables: "
        + str(missing)
        + "\n\nThe notebook must already contain X and y."
    )

# Try to locate subject vector automatically
SUBJECT_VAR_CANDIDATES = [
    "subjects",
    "subject",
    "subject_ids",
    "subject_id",
    "subject_labels",
    "groups",
    "groups_loso"
]

SUBJECT_VAR = None

for candidate in SUBJECT_VAR_CANDIDATES:
    if candidate in globals():
        SUBJECT_VAR = candidate
        break

if SUBJECT_VAR is None:
    raise RuntimeError(
        "Could not find the subject-ID vector.\n"
        "Expected one of:\n"
        + ", ".join(SUBJECT_VAR_CANDIDATES)
    )

subjects_raw = globals()[SUBJECT_VAR]

# =============================================================================
# 3. CONVERT DATA
# =============================================================================

X = np.asarray(X, dtype=np.float32)
y = np.asarray(y).astype(np.int64)
subjects_raw = np.asarray(subjects_raw).astype(str)

print("\n[1] DATASET")
print("X shape      :", X.shape)
print("X dtype      :", X.dtype)
print("y shape      :", y.shape)
print("Subjects     :", len(np.unique(subjects_raw)))
print("Classes      :", np.unique(y))
print("Class counts :", np.bincount(y))

# =============================================================================
# 4. VALIDATION
# =============================================================================

if X.ndim != 3:
    raise ValueError(
        f"Expected X to be 3-D, got {X.ndim}-D"
    )

if X.shape[1] != EXPECTED_CHANNELS:
    raise ValueError(
        f"Expected {EXPECTED_CHANNELS} channels, "
        f"got {X.shape[1]}"
    )

if X.shape[2] != EXPECTED_SAMPLES:
    raise ValueError(
        f"Expected {EXPECTED_SAMPLES} samples, "
        f"got {X.shape[2]}"
    )

if len(X) != len(y):
    raise ValueError("X and y length mismatch.")

if len(X) != len(subjects_raw):
    raise ValueError(
        "X and subject vector length mismatch."
    )

if np.isnan(X).any():
    raise ValueError("NaNs detected.")

if np.isinf(X).any():
    raise ValueError("Infs detected.")

classes = np.unique(y)

if len(classes) != NUM_CLASSES:
    raise ValueError(
        f"Expected {NUM_CLASSES} classes, "
        f"found {len(classes)}: {classes}"
    )

# Re-map labels to 0...(C-1)
unique_classes = np.sort(classes)

class_mapping = {
    int(original): int(i)
    for i, original in enumerate(unique_classes)
}

y = np.asarray(
    [class_mapping[int(v)] for v in y],
    dtype=np.int64
)

print("\nDataset validation: PASS")

# =============================================================================
# 5. SUBJECT LABEL NORMALIZATION
# =============================================================================

def canonical_subject(raw_subject):
    """
    Convert subject IDs to a stable string.

    Examples:
        BCI-IV-2a::S01 -> BCI-IV-2a::S01
        1              -> 1
    """

    s = str(raw_subject).strip()

    return s


subjects = np.asarray(
    [canonical_subject(s) for s in subjects_raw]
)

unique_subjects = np.unique(subjects)

print("\nUnique LOSO subjects:", len(unique_subjects))

# =============================================================================
# 6. NORMALIZATION
# =============================================================================

def normalize_trials_numpy(X_in, clip_value=8.0):
    """
    Trial-wise normalization.

    Each:
        trial x channel

    is standardized across time.

    This removes large absolute-amplitude differences between trials
    while preserving temporal/spatial structure.
    """

    Xn = X_in.astype(np.float32, copy=True)

    if not USE_TRIAL_NORMALIZATION:
        return Xn

    mean = Xn.mean(
        axis=2,
        keepdims=True
    )

    std = Xn.std(
        axis=2,
        keepdims=True
    )

    Xn = (
        Xn - mean
    ) / (
        std + 1e-6
    )

    Xn = np.nan_to_num(
        Xn,
        nan=0.0,
        posinf=clip_value,
        neginf=-clip_value
    )

    Xn = np.clip(
        Xn,
        -clip_value,
        clip_value
    )

    return Xn.astype(np.float32)


# =============================================================================
# 7. BALANCED CLASS WEIGHTS
# =============================================================================

def create_class_weights(labels):

    counts = np.bincount(
        labels,
        minlength=NUM_CLASSES
    ).astype(np.float64)

    counts[counts == 0] = 1.0

    weights = (
        len(labels)
        /
        (
            NUM_CLASSES * counts
        )
    )

    return weights.astype(np.float32)


# =============================================================================
# 8. CONVORELENET
# =============================================================================

class TemporalConvBranch(nn.Module):

    def __init__(
        self,
        n_channels=22,
        dropout=0.40
    ):

        super().__init__()

        # Multiple temporal scales
        self.conv3 = nn.Sequential(

            nn.Conv2d(
                1,
                16,
                kernel_size=(1, 3),
                padding=(0, 1),
                bias=False
            ),

            nn.GroupNorm(
                4,
                16
            ),

            nn.GELU()
        )

        self.conv5 = nn.Sequential(

            nn.Conv2d(
                1,
                16,
                kernel_size=(1, 5),
                padding=(0, 2),
                bias=False
            ),

            nn.GroupNorm(
                4,
                16
            ),

            nn.GELU()
        )

        self.conv7 = nn.Sequential(

            nn.Conv2d(
                1,
                16,
                kernel_size=(1, 7),
                padding=(0, 3),
                bias=False
            ),

            nn.GroupNorm(
                4,
                16
            ),

            nn.GELU()
        )

        # Fusion
        self.fusion = nn.Sequential(

            nn.Conv2d(
                48,
                32,
                kernel_size=(1, 3),
                padding=(0, 1),
                bias=False
            ),

            nn.GroupNorm(
                8,
                32
            ),

            nn.GELU(),

            nn.Conv2d(
                32,
                64,
                kernel_size=(1, 3),
                padding=(0, 1),
                bias=False
            ),

            nn.GroupNorm(
                8,
                64
            ),

            nn.GELU(),

            nn.Dropout2d(
                dropout
            )
        )

        # Spatial compression across electrodes
        self.spatial = nn.Sequential(

            nn.Conv2d(
                64,
                64,
                kernel_size=(n_channels, 1),
                bias=False
            ),

            nn.GroupNorm(
                8,
                64
            ),

            nn.GELU(),

            nn.Dropout2d(
                dropout
            )
        )

        self.pool = nn.AvgPool2d(
            kernel_size=(1, 8),
            stride=(1, 8)
        )

    def forward(self, x):

        a = self.conv3(x)
        b = self.conv5(x)
        c = self.conv7(x)

        x = torch.cat(
            [a, b, c],
            dim=1
        )

        x = self.fusion(x)

        x = self.spatial(x)

        x = self.pool(x)

        return x


class SpatialConvBranch(nn.Module):

    def __init__(
        self,
        n_channels=22,
        dropout=0.40
    ):

        super().__init__()

        # First extract temporal information
        self.temporal = nn.Sequential(

            nn.Conv2d(
                1,
                16,
                kernel_size=(1, 7),
                padding=(0, 3),
                bias=False
            ),

            nn.GroupNorm(
                4,
                16
            ),

            nn.GELU()
        )

        # Spatial convolution over electrodes
        self.spatial1 = nn.Sequential(

            nn.Conv2d(
                16,
                32,
                kernel_size=(n_channels, 1),
                bias=False
            ),

            nn.GroupNorm(
                8,
                32
            ),

            nn.GELU(),

            nn.Dropout2d(
                dropout
            )
        )

        self.spatial2 = nn.Sequential(

            nn.Conv2d(
                32,
                64,
                kernel_size=(1, 3),
                padding=(0, 1),
                bias=False
            ),

            nn.GroupNorm(
                8,
                64
            ),

            nn.GELU(),

            nn.Dropout2d(
                dropout
            )
        )

        self.pool = nn.AvgPool2d(
            kernel_size=(1, 8),
            stride=(1, 8)
        )

    def forward(self, x):

        x = self.temporal(x)

        x = self.spatial1(x)

        x = self.spatial2(x)

        x = self.pool(x)

        return x


class PositionalEncoding(nn.Module):

    def __init__(
        self,
        d_model,
        max_len=512,
        dropout=0.10
    ):

        super().__init__()

        self.dropout = nn.Dropout(
            dropout
        )

        pe = torch.zeros(
            max_len,
            d_model
        )

        position = torch.arange(
            0,
            max_len,
            dtype=torch.float32
        ).unsqueeze(1)

        div_term = torch.exp(
            torch.arange(
                0,
                d_model,
                2,
                dtype=torch.float32
            )
            *
            (
                -math.log(10000.0)
                / d_model
            )
        )

        pe[:, 0::2] = torch.sin(
            position * div_term
        )

        pe[:, 1::2] = torch.cos(
            position * div_term
        )

        pe = pe.unsqueeze(0)

        self.register_buffer(
            "pe",
            pe
        )

    def forward(self, x):

        x = x + self.pe[:, :x.size(1)]

        return self.dropout(x)


class ConvoReLeNet(nn.Module):

    def __init__(
        self,
        n_channels=22,
        n_samples=640,
        n_classes=3
    ):

        super().__init__()

        self.temporal_branch = TemporalConvBranch(
            n_channels=n_channels,
            dropout=CONV_DROPOUT
        )

        self.spatial_branch = SpatialConvBranch(
            n_channels=n_channels,
            dropout=CONV_DROPOUT
        )

        # Two branches => 64 + 64 = 128
        self.fusion = nn.Sequential(

            nn.Conv1d(
                128,
                D_MODEL,
                kernel_size=1,
                bias=False
            ),

            nn.GroupNorm(
                16,
                D_MODEL
            ),

            nn.GELU()
        )

        self.positional = PositionalEncoding(
            d_model=D_MODEL,
            max_len=256,
            dropout=0.10
        )

        encoder_layer = nn.TransformerEncoderLayer(

            d_model=D_MODEL,

            nhead=N_HEADS,

            dim_feedforward=4 * D_MODEL,

            dropout=0.10,

            activation="gelu",

            batch_first=True,

            norm_first=True
        )

        self.transformer = nn.TransformerEncoder(

            encoder_layer,

            num_layers=N_TRANSFORMER_LAYERS,

            norm=nn.LayerNorm(
                D_MODEL
            )
        )

        self.bilstm = nn.LSTM(

            input_size=D_MODEL,

            hidden_size=LSTM_HIDDEN,

            num_layers=N_LSTM_LAYERS,

            batch_first=True,

            bidirectional=True,

            dropout=LSTM_DROPOUT
            if N_LSTM_LAYERS > 1
            else 0.0
        )

        self.classifier = nn.Sequential(

            nn.LayerNorm(
                2 * LSTM_HIDDEN
            ),

            nn.Linear(
                2 * LSTM_HIDDEN,
                256
            ),

            nn.GELU(),

            nn.Dropout(
                MLP_DROPOUT
            ),

            nn.Linear(
                256,
                128
            ),

            nn.GELU(),

            nn.Dropout(
                MLP_DROPOUT
            ),

            nn.Linear(
                128,
                n_classes
            )
        )

    def forward(
        self,
        x,
        return_features=False
    ):

        # -----------------------------------------------------
        # x:
        #   B x 1 x 22 x 640
        # -----------------------------------------------------

        temporal = self.temporal_branch(x)

        spatial = self.spatial_branch(x)

        # Both:
        #   B x 64 x 1 x T
        # Concatenate:
        #   B x 128 x 1 x T

        x = torch.cat(
            [temporal, spatial],
            dim=1
        )

        x = x.squeeze(2)

        # B x 128 x T
        x = self.fusion(x)

        # B x T x 128
        x = x.transpose(
            1,
            2
        )

        x = self.positional(x)

        x = self.transformer(x)

        x, _ = self.bilstm(x)

        # Global temporal pooling
        features = x.mean(
            dim=1
        )

        logits = self.classifier(
            features
        )

        if return_features:
            return logits, features

        return logits


# =============================================================================
# 9. PARAMETER REPORT
# =============================================================================

def report_parameters(model):

    total = sum(
        p.numel()
        for p in model.parameters()
    )

    trainable = sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

    print(
        f"Total parameters    : {total:,}"
    )

    print(
        f"Trainable parameters: {trainable:,}"
    )

    return total, trainable


# =============================================================================
# 10. TARGET SPLITTING
# =============================================================================

def stratified_target_split(
    X_target,
    y_target,
    calibration_ratio=0.20,
    seed=42
):

    rng = np.random.default_rng(seed)

    calibration_idx = []
    test_idx = []

    for cls in np.unique(y_target):

        idx = np.where(
            y_target == cls
        )[0]

        idx = idx.copy()

        rng.shuffle(idx)

        n_cal = max(
            1,
            int(
                round(
                    len(idx)
                    * calibration_ratio
                )
            )
        )

        calibration_idx.extend(
            idx[:n_cal]
        )

        test_idx.extend(
            idx[n_cal:]
        )

    calibration_idx = np.asarray(
        calibration_idx,
        dtype=np.int64
    )

    test_idx = np.asarray(
        test_idx,
        dtype=np.int64
    )

    rng.shuffle(
        calibration_idx
    )

    rng.shuffle(
        test_idx
    )

    return calibration_idx, test_idx


# =============================================================================
# 11. SOURCE DATA PREPARATION
# =============================================================================

def prepare_source_data(
    X_source,
    y_source
):

    print(
        "\n    Normalizing source trials..."
    )

    X_source = normalize_trials_numpy(
        X_source
    )

    # Add image dimension
    X_source = (
        X_source
        .astype(np.float32)
    )

    X_source = X_source[:, None, :, :]

    return X_source, y_source


# =============================================================================
# 12. DATA LOADER
# =============================================================================

def make_balanced_loader(
    X_np,
    y_np,
    batch_size,
    shuffle=True
):

    Xt = torch.from_numpy(
        X_np
    ).float()

    yt = torch.from_numpy(
        y_np
    ).long()

    ds = TensorDataset(
        Xt,
        yt
    )

    if shuffle:

        class_weights = create_class_weights(
            y_np
        )

        sample_weights = (
            class_weights[y_np]
        )

        sampler = WeightedRandomSampler(

            weights=torch.as_tensor(
                sample_weights,
                dtype=torch.double
            ),

            num_samples=len(
                sample_weights
            ),

            replacement=True
        )

        return DataLoader(
            ds,
            batch_size=batch_size,
            sampler=sampler,
            num_workers=0,
            pin_memory=False
        )

    return DataLoader(
        ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=False
    )


# =============================================================================
# 13. EVALUATION
# =============================================================================

@torch.no_grad()
def evaluate_model(
    model,
    X_np,
    y_np,
    batch_size=64
):

    model.eval()

    Xt = torch.from_numpy(
        X_np
    ).float()

    yt = torch.from_numpy(
        y_np
    ).long()

    ds = TensorDataset(
        Xt,
        yt
    )

    loader = DataLoader(
        ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0
    )

    predictions = []
    targets = []

    losses = []

    criterion = nn.CrossEntropyLoss()

    for xb, yb in loader:

        xb = xb.to(
            DEVICE
        )

        yb = yb.to(
            DEVICE
        )

        logits = model(
            xb
        )

        loss = criterion(
            logits,
            yb
        )

        pred = logits.argmax(
            dim=1
        )

        predictions.extend(
            pred.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        targets.extend(
            yb.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        losses.append(
            loss.item()
        )

    predictions = np.asarray(
        predictions
    )

    targets = np.asarray(
        targets
    )

    acc = accuracy_score(
        targets,
        predictions
    )

    f1 = f1_score(
        targets,
        predictions,
        average="macro",
        zero_division=0
    )

    kappa = cohen_kappa_score(
        targets,
        predictions
    )

    cm = confusion_matrix(
        targets,
        predictions,
        labels=np.arange(
            NUM_CLASSES
        )
    )

    return {
        "loss": float(
            np.mean(losses)
        ),
        "accuracy": float(acc),
        "macro_f1": float(f1),
        "kappa": float(kappa),
        "confusion_matrix": cm,
        "predictions": predictions,
        "targets": targets
    }


# =============================================================================
# 14. TRAINING FUNCTION
# =============================================================================

def train_model(
    model,
    train_X,
    train_y,
    epochs,
    batch_size,
    lr,
    weight_decay,
    stage_name,
    scheduler_type="cosine"
):

    loader = make_balanced_loader(
        train_X,
        train_y,
        batch_size=batch_size,
        shuffle=True
    )

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.AdamW(

        model.parameters(),

        lr=lr,

        weight_decay=weight_decay,

        betas=(0.9, 0.999)
    )

    # Warm-up followed by cosine decay
    warmup_epochs = min(
        10,
        max(1, epochs // 10)
    )

    def lr_lambda(epoch):

        if epoch < warmup_epochs:

            return float(
                epoch + 1
            ) / float(
                warmup_epochs
            )

        progress = (
            epoch - warmup_epochs
        ) / max(
            1,
            epochs - warmup_epochs
        )

        return 0.5 * (
            1
            +
            math.cos(
                math.pi
                * progress
            )
        )

    scheduler = torch.optim.lr_scheduler.LambdaLR(
        optimizer,
        lr_lambda
    )

    history = []

    best_state = None

    best_acc = -np.inf

    best_loss = np.inf

    for epoch in range(
        1,
        epochs + 1
    ):

        model.train()

        total_loss = 0.0
        total = 0
        correct = 0

        for xb, yb in loader:

            xb = xb.to(
                DEVICE
            )

            yb = yb.to(
                DEVICE
            )

            optimizer.zero_grad(
                set_to_none=True
            )

            logits = model(
                xb
            )

            loss = criterion(
                logits,
                yb
            )

            if not torch.isfinite(
                loss
            ):

                raise RuntimeError(
                    f"Non-finite loss "
                    f"at epoch {epoch}"
                )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                GRAD_CLIP
            )

            optimizer.step()

            total_loss += (
                loss.item()
                * xb.size(0)
            )

            pred = logits.argmax(
                dim=1
            )

            correct += (
                pred == yb
            ).sum().item()

            total += yb.size(0)

        scheduler.step()

        epoch_loss = (
            total_loss
            /
            max(
                1,
                total
            )
        )

        epoch_acc = (
            100.0
            *
            correct
            /
            max(
                1,
                total
            )
        )

        current_lr = (
            optimizer.param_groups[0]["lr"]
        )

        history.append({
            "epoch": epoch,
            "loss": epoch_loss,
            "accuracy": epoch_acc,
            "lr": current_lr
        })

        # Save best training checkpoint
        if (
            epoch_acc > best_acc
            or (
                epoch_acc == best_acc
                and epoch_loss < best_loss
            )
        ):

            best_acc = epoch_acc

            best_loss = epoch_loss

            best_state = copy.deepcopy(
                model.state_dict()
            )

        if (
            epoch == 1
            or epoch % 10 == 0
            or epoch == epochs
        ):

            print(
                f"{stage_name:<15}"
                f"{epoch:4d}/{epochs:<4d} | "
                f"loss={epoch_loss:.4f} | "
                f"acc={epoch_acc:6.2f}% | "
                f"lr={current_lr:.2e}"
            )

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    return model, pd.DataFrame(
        history
    )


# =============================================================================
# 15. MICRO OVERFIT TEST
# =============================================================================

def run_micro_overfit():

    print("\n" + "=" * 80)
    print("MICRO OVERFIT TEST")
    print("=" * 80)

    rng = np.random.default_rng(
        SEED
    )

    selected = []

    for cls in np.unique(y):

        cls_idx = np.where(
            y == cls
        )[0]

        n = min(
            MICRO_OVERFIT_N_PER_CLASS,
            len(cls_idx)
        )

        selected.extend(
            rng.choice(
                cls_idx,
                size=n,
                replace=False
            ).tolist()
        )

    selected = np.asarray(
        selected,
        dtype=np.int64
    )

    X_micro = X[selected]

    y_micro = y[selected]

    X_micro = normalize_trials_numpy(
        X_micro
    )

    X_micro = X_micro[:, None, :, :]

    print(
        "Micro dataset:",
        X_micro.shape
    )

    print(
        "Class distribution:",
        np.bincount(
            y_micro,
            minlength=NUM_CLASSES
        )
    )

    model = ConvoReLeNet(
        n_channels=EXPECTED_CHANNELS,
        n_samples=EXPECTED_SAMPLES,
        n_classes=NUM_CLASSES
    ).to(
        DEVICE
    )

    report_parameters(
        model
    )

    model, history = train_model(

        model=model,

        train_X=X_micro,

        train_y=y_micro,

        epochs=MICRO_OVERFIT_EPOCHS,

        batch_size=min(
            8,
            len(X_micro)
        ),

        lr=1e-3,

        weight_decay=0.0,

        stage_name="micro-fit"
    )

    result = evaluate_model(
        model,
        X_micro,
        y_micro,
        batch_size=8
    )

    print(
        "\nMICRO OVERFIT ACCURACY:",
        f"{100 * result['accuracy']:.2f}%"
    )

    if result["accuracy"] >= 0.95:

        print(
            "STATUS: PASS"
        )

        print(
            "The architecture and "
            "gradient path can learn the data."
        )

    elif result["accuracy"] >= 0.70:

        print(
            "STATUS: PARTIAL"
        )

        print(
            "The architecture learns, "
            "but optimization is weak."
        )

    else:

        print(
            "STATUS: FAIL"
        )

        print(
            "The model cannot memorize "
            "a tiny dataset."
        )

    return model, history, result


# =============================================================================
# 16. RUN MICRO TEST BEFORE LOSO
# =============================================================================

if RUN_MICRO_OVERFIT_TEST:

    micro_model, micro_history, micro_result = (
        run_micro_overfit()
    )

    del micro_model
    del micro_history
    del micro_result

    gc.collect()

    if DEVICE.type == "mps":

        try:
            torch.mps.empty_cache()
        except Exception:
            pass


# =============================================================================
# 17. SELECT TARGETS
# =============================================================================

if RUN_FULL_LOSO:

    targets_to_run = list(
        unique_subjects
    )

else:

    if TARGET_SUBJECTS is None:

        targets_to_run = [
            unique_subjects[0]
        ]

    else:

        targets_to_run = [
            str(s)
            for s in TARGET_SUBJECTS
        ]

# Validate requested targets
for target in targets_to_run:

    if target not in set(
        unique_subjects
    ):

        raise ValueError(
            f"Target subject '{target}' "
            f"not found."
        )

print("\n" + "=" * 80)
print("TARGETS SCHEDULED")
print("=" * 80)

for t in targets_to_run:
    print(t)


# =============================================================================
# 18. LOSO PIPELINE
# =============================================================================

all_fold_results = []
all_confusions = []
all_histories = []

global_start = time.time()

print("\n" + "=" * 80)
print("STARTING CONVORELENET LOSO")
print("=" * 80)

for fold_number, target_subject in enumerate(
    targets_to_run,
    start=1
):

    fold_start = time.time()

    print("\n\n")
    print("=" * 80)
    print(
        f"FOLD {fold_number}/{len(targets_to_run)}"
    )
    print("=" * 80)
    print(
        f"TARGET SUBJECT: {target_subject}"
    )
    print("=" * 80)

    # -------------------------------------------------------------------------
    # Target mask
    # -------------------------------------------------------------------------

    target_mask = (
        subjects == target_subject
    )

    source_mask = ~target_mask

    X_source_raw = X[
        source_mask
    ]

    y_source = y[
        source_mask
    ]

    X_target_raw = X[
        target_mask
    ]

    y_target = y[
        target_mask
    ]

    print(
        "\nSource trials:",
        len(X_source_raw)
    )

    print(
        "Target trials:",
        len(X_target_raw)
    )

    if len(X_target_raw) < 3:

        print(
            "Skipping target due to "
            "insufficient trials."
        )

        continue

    # -------------------------------------------------------------------------
    # Split target into calibration and held-out test
    # -------------------------------------------------------------------------

    cal_idx, test_idx = (
        stratified_target_split(
            X_target_raw,
            y_target,
            calibration_ratio=TARGET_CALIBRATION_RATIO,
            seed=SEED + fold_number
        )
    )

    X_cal_raw = X_target_raw[
        cal_idx
    ]

    y_cal = y_target[
        cal_idx
    ]

    X_test_raw = X_target_raw[
        test_idx
    ]

    y_test = y_target[
        test_idx
    ]

    print(
        "Target calibration:",
        len(X_cal_raw)
    )

    print(
        "Target test:",
        len(X_test_raw)
    )

    print(
        "\nCalibration classes:",
        np.bincount(
            y_cal,
            minlength=NUM_CLASSES
        )
    )

    print(
        "Test classes:",
        np.bincount(
            y_test,
            minlength=NUM_CLASSES
        )
    )

    # -------------------------------------------------------------------------
    # IMPORTANT:
    # Normalize every trial independently.
    # No test labels are used for training.
    # -------------------------------------------------------------------------

    print(
        "\nPreparing source data..."
    )

    X_source, y_source = (
        prepare_source_data(
            X_source_raw,
            y_source
        )
    )

    print(
        "Preparing target calibration..."
    )

    X_cal = normalize_trials_numpy(
        X_cal_raw
    )

    X_cal = X_cal[:, None, :, :]

    print(
        "Preparing target test..."
    )

    X_test = normalize_trials_numpy(
        X_test_raw
    )

    X_test = X_test[:, None, :, :]

    # -------------------------------------------------------------------------
    # Create model
    # -------------------------------------------------------------------------

    print(
        "\nCreating ConvoReLeNet..."
    )

    model = ConvoReLeNet(
        n_channels=EXPECTED_CHANNELS,
        n_samples=EXPECTED_SAMPLES,
        n_classes=NUM_CLASSES
    ).to(
        DEVICE
    )

    report_parameters(
        model
    )

    # -------------------------------------------------------------------------
    # SOURCE PRETRAINING
    # -------------------------------------------------------------------------

    print(
        "\n" + "-" * 80
    )

    print(
        "--- SOURCE PRETRAINING ---"
    )

    print(
        "-" * 80
    )

    source_start = time.time()

    model, pretrain_history = train_model(

        model=model,

        train_X=X_source,

        train_y=y_source,

        epochs=PRETRAIN_EPOCHS,

        batch_size=PRETRAIN_BATCH_SIZE,

        lr=PRETRAIN_LR,

        weight_decay=PRETRAIN_WEIGHT_DECAY,

        stage_name="pretrain"
    )

    source_time = (
        time.time()
        - source_start
    )

    print(
        f"\nSource pretraining time: "
        f"{source_time / 60:.2f} min"
    )

    # -------------------------------------------------------------------------
    # SOURCE -> TARGET BEFORE ADAPTATION
    # -------------------------------------------------------------------------

    print(
        "\n--- SOURCE → TARGET BEFORE FINE-TUNING ---"
    )

    source_cal_result = evaluate_model(
        model,
        X_cal,
        y_cal,
        batch_size=32
    )

    source_test_result = evaluate_model(
        model,
        X_test,
        y_test,
        batch_size=32
    )

    print(
        f"Source → target calibration: "
        f"{100 * source_cal_result['accuracy']:.2f}%"
    )

    print(
        f"Source → target test       : "
        f"{100 * source_test_result['accuracy']:.2f}%"
    )

    # -------------------------------------------------------------------------
    # TARGET FINE-TUNING
    #
    # Paper protocol:
    # convolutional and recurrent layers can adapt;
    # Transformer layers are kept partially frozen.
    # -------------------------------------------------------------------------

    print(
        "\n" + "-" * 80
    )

    print(
        "--- CONFIGURING CONSERVATIVE TRANSFER ---"
    )

    print(
        "-" * 80
    )

    # Freeze Transformer encoder only.
    # Convolutional branches, fusion, positional encoding,
    # BiLSTM and classifier remain trainable.

    for name, parameter in model.named_parameters():

        if (
            name.startswith(
                "transformer."
            )
        ):

            parameter.requires_grad = False

        else:

            parameter.requires_grad = True

    trainable = sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

    total = sum(
        p.numel()
        for p in model.parameters()
    )

    print(
        "Trainable parameters:",
        f"{trainable:,}"
    )

    print(
        "Total parameters    :",
        f"{total:,}"
    )

    # -------------------------------------------------------------------------
    # TARGET FINE-TUNING
    # -------------------------------------------------------------------------

    print(
        "\n--- TARGET FINE-TUNING ---"
    )

    ft_start = time.time()

    model, target_ft_history = train_model(

        model=model,

        train_X=X_cal,

        train_y=y_cal,

        epochs=TARGET_FT_EPOCHS,

        batch_size=TARGET_FT_BATCH_SIZE,

        lr=TARGET_FT_LR,

        weight_decay=TARGET_FT_WEIGHT_DECAY,

        stage_name="target-ft"
    )

    ft_time = (
        time.time()
        - ft_start
    )

    print(
        f"\nTarget fine-tuning time: "
        f"{ft_time / 60:.2f} min"
    )

    # -------------------------------------------------------------------------
    # TARGET CALIBRATION RESULT
    # -------------------------------------------------------------------------

    print(
        "\n--- ADAPTED MODEL / CALIBRATION ---"
    )

    cal_result = evaluate_model(
        model,
        X_cal,
        y_cal,
        batch_size=16
    )

    print(
        f"Calibration accuracy: "
        f"{100 * cal_result['accuracy']:.2f}%"
    )

    print(
        f"Calibration macro-F1: "
        f"{cal_result['macro_f1']:.4f}"
    )

    print(
        f"Calibration kappa   : "
        f"{cal_result['kappa']:.4f}"
    )

    print(
        "\nCalibration confusion matrix:"
    )

    print(
        cal_result[
            "confusion_matrix"
        ]
    )

    # -------------------------------------------------------------------------
    # COMPLETELY HELD-OUT TARGET TEST
    # -------------------------------------------------------------------------

    print(
        "\n--- COMPLETELY HELD-OUT TARGET TEST ---"
    )

    test_result = evaluate_model(
        model,
        X_test,
        y_test,
        batch_size=32
    )

    print(
        "\n"
    )

    print(
        f"TARGET TEST ACCURACY : "
        f"{100 * test_result['accuracy']:.2f}%"
    )

    print(
        f"TARGET TEST MACRO-F1 : "
        f"{test_result['macro_f1']:.4f}"
    )

    print(
        f"TARGET TEST KAPPA    : "
        f"{test_result['kappa']:.4f}"
    )

    print(
        "\nTarget test confusion matrix:"
    )

    print(
        test_result[
            "confusion_matrix"
        ]
    )

    # -------------------------------------------------------------------------
    # FOLD RESULT
    # -------------------------------------------------------------------------

    fold_time = (
        time.time()
        - fold_start
    )

    fold_result = {

        "fold": fold_number,

        "target_subject": target_subject,

        "source_trials": len(
            X_source
        ),

        "target_trials": len(
            X_target_raw
        ),

        "calibration_trials": len(
            X_cal
        ),

        "test_trials": len(
            X_test
        ),

        "source_cal_accuracy": (
            100
            *
            source_cal_result[
                "accuracy"
            ]
        ),

        "source_test_accuracy": (
            100
            *
            source_test_result[
                "accuracy"
            ]
        ),

        "calibration_accuracy": (
            100
            *
            cal_result[
                "accuracy"
            ]
        ),

        "calibration_macro_f1": (
            cal_result[
                "macro_f1"
            ]
        ),

        "calibration_kappa": (
            cal_result[
                "kappa"
            ]
        ),

        "target_test_accuracy": (
            100
            *
            test_result[
                "accuracy"
            ]
        ),

        "target_test_macro_f1": (
            test_result[
                "macro_f1"
            ]
        ),

        "target_test_kappa": (
            test_result[
                "kappa"
            ]
        ),

        "pretrain_best_accuracy": (
            float(
                pretrain_history[
                    "accuracy"
                ].max()
            )
        ),

        "target_ft_best_accuracy": (
            float(
                target_ft_history[
                    "accuracy"
                ].max()
            )
        ),

        "fold_minutes": (
            fold_time / 60.0
        )
    }

    all_fold_results.append(
        fold_result
    )

    all_confusions.append({
        "target_subject": target_subject,
        "confusion_matrix": test_result[
            "confusion_matrix"
        ]
    })

    h1 = pretrain_history.copy()
    h1["stage"] = "pretrain"
    h1["target_subject"] = target_subject

    h2 = target_ft_history.copy()
    h2["stage"] = "target-ft"
    h2["target_subject"] = target_subject

    all_histories.append(h1)
    all_histories.append(h2)

    # -------------------------------------------------------------------------
    # SAVE INTERMEDIATE RESULTS
    # -------------------------------------------------------------------------

    fold_df = pd.DataFrame(
        all_fold_results
    )

    history_df = pd.concat(
        all_histories,
        ignore_index=True
    )

    # Save in notebook working directory
    fold_df.to_csv(
        "convor elenet_module5_fold_results.csv".replace(
            " ",
            ""
        ),
        index=False
    )

    history_df.to_csv(
        "convorelenet_module5_training_history.csv",
        index=False
    )

    # -------------------------------------------------------------------------
    # MEMORY CLEANUP
    # -------------------------------------------------------------------------

    del model

    del X_source
    del X_cal
    del X_test

    del X_source_raw
    del X_cal_raw
    del X_test_raw

    gc.collect()

    if DEVICE.type == "mps":

        try:
            torch.mps.empty_cache()
        except Exception:
            pass

    print(
        f"\nFold completed in "
        f"{fold_time / 60:.2f} minutes."
    )


# =============================================================================
# 19. FINAL RESULTS
# =============================================================================

print("\n\n")
print("=" * 80)
print("CONVORELENET — FINAL RESULTS")
print("=" * 80)

if len(all_fold_results) == 0:

    print(
        "No folds were completed."
    )

else:

    results_df = pd.DataFrame(
        all_fold_results
    )

    print(
        results_df.to_string(
            index=False
        )
    )

    print(
        "\n" + "-" * 80
    )

    print(
        "LOSO SUMMARY"
    )

    print(
        "-" * 80
    )

    mean_acc = (
        results_df[
            "target_test_accuracy"
        ].mean()
    )

    std_acc = (
        results_df[
            "target_test_accuracy"
        ].std(
            ddof=1
        )
        if len(results_df) > 1
        else 0.0
    )

    mean_f1 = (
        results_df[
            "target_test_macro_f1"
        ].mean()
    )

    std_f1 = (
        results_df[
            "target_test_macro_f1"
        ].std(
            ddof=1
        )
        if len(results_df) > 1
        else 0.0
    )

    mean_kappa = (
        results_df[
            "target_test_kappa"
        ].mean()
    )

    std_kappa = (
        results_df[
            "target_test_kappa"
        ].std(
            ddof=1
        )
        if len(results_df) > 1
        else 0.0
    )

    print(
        f"\nMean target-test accuracy : "
        f"{mean_acc:.2f} ± {std_acc:.2f}%"
    )

    print(
        f"Mean target-test macro-F1 : "
        f"{mean_f1:.4f} ± {std_f1:.4f}"
    )

    print(
        f"Mean target-test kappa    : "
        f"{mean_kappa:.4f} ± {std_kappa:.4f}"
    )

    print(
        "\nChance level:",
        f"{100 / NUM_CLASSES:.2f}%"
    )

    # ---------------------------------------------------------
    # Save final results
    # ---------------------------------------------------------

    results_df.to_csv(
        "convorelenet_module5_FINAL_RESULTS.csv",
        index=False
    )

    if len(all_histories) > 0:

        history_df = pd.concat(
            all_histories,
            ignore_index=True
        )

        history_df.to_csv(
            "convorelenet_module5_FINAL_HISTORY.csv",
            index=False
        )

    print(
        "\nSaved:"
    )

    print(
        "  convorelenet_module5_FINAL_RESULTS.csv"
    )

    print(
        "  convorelenet_module5_FINAL_HISTORY.csv"
    )


# =============================================================================
# 20. TOTAL RUNTIME
# =============================================================================

total_runtime = (
    time.time()
    - global_start
)

print(
    "\nTotal runtime:",
    f"{total_runtime / 60:.2f} minutes"
)

print("\n" + "=" * 80)
print("PIPELINE COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# MODULE 5 — CONVORELENET MICRO OVERFIT TEST
# ============================================================

import copy
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

print("\n" + "=" * 80)
print("CONVORELENET MICRO OVERFIT TEST")
print("=" * 80)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------
OVERFIT_N = 48
EPOCHS = 150
BATCH_SIZE = 16
LR = 1e-3

device = torch.device(
    "mps" if torch.backends.mps.is_available()
    else "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)

# ------------------------------------------------------------
# Select a balanced 48-sample subset
# ------------------------------------------------------------
X_np = np.asarray(X, dtype=np.float32)
y_np = np.asarray(y, dtype=np.int64)

selected = []

rng = np.random.default_rng(1234)

for cls in np.unique(y_np):

    cls_idx = np.where(y_np == cls)[0]

    take = min(OVERFIT_N // 3, len(cls_idx))

    selected.extend(
        rng.choice(
            cls_idx,
            size=take,
            replace=False
        ).tolist()
    )

selected = np.asarray(selected)

X_small = X_np[selected]
y_small = y_np[selected]

print("\nSelected samples:", len(X_small))
print("Shape:", X_small.shape)

print(
    "Class counts:",
    dict(zip(*np.unique(y_small, return_counts=True)))
)

# ------------------------------------------------------------
# VERY IMPORTANT:
# standardize each trial independently
# ------------------------------------------------------------
X_small = (
    X_small -
    X_small.mean(axis=2, keepdims=True)
)

X_small = (
    X_small /
    (
        X_small.std(axis=2, keepdims=True)
        + 1e-8
    )
)

print(
    "\nAfter trial normalization:",
    "mean=",
    float(X_small.mean()),
    "std=",
    float(X_small.std())
)

# ------------------------------------------------------------
# Tensor format expected by ConvNet
# ------------------------------------------------------------
X_t = torch.from_numpy(
    X_small
).float().unsqueeze(1)

y_t = torch.from_numpy(
    y_small
).long()

print("Tensor shape:", tuple(X_t.shape))
print("Label shape :", tuple(y_t.shape))

# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------
ds = TensorDataset(X_t, y_t)

loader = DataLoader(
    ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    drop_last=False
)

# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------
model = ConvoReLeNet(
    in_channels=1,
    n_classes=3,
    n_electrodes=22,
    n_samples=640
).to(device)

print("\nModel created.")

# ------------------------------------------------------------
# Train
# ------------------------------------------------------------
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LR,
    weight_decay=0.0
)

best_acc = 0.0

for epoch in range(1, EPOCHS + 1):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for xb, yb in loader:

        xb = xb.to(device)
        yb = yb.to(device)

        optimizer.zero_grad(set_to_none=True)

        logits = model(xb)

        # If model returns tuple/list, use first item
        if isinstance(logits, (tuple, list)):
            logits = logits[0]

        loss = criterion(logits, yb)

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item() * xb.size(0)
        )

        pred = logits.argmax(dim=1)

        correct += (
            pred == yb
        ).sum().item()

        total += xb.size(0)

    epoch_loss = running_loss / total
    epoch_acc = 100.0 * correct / total

    best_acc = max(
        best_acc,
        epoch_acc
    )

    if (
        epoch == 1
        or epoch % 10 == 0
        or epoch_acc >= 95.0
    ):
        print(
            f"epoch {epoch:3d}/{EPOCHS} | "
            f"loss={epoch_loss:.4f} | "
            f"acc={epoch_acc:6.2f}%"
        )

    if epoch_acc >= 99.0:
        print(
            "\nReached ≥99% training accuracy."
        )
        break

print("\n" + "=" * 80)
print("MICRO OVERFIT RESULT")
print("=" * 80)

print(
    f"Best training accuracy: {best_acc:.2f}%"
)

if best_acc >= 95.0:
    print(
        "\nPASS:"
        "\nThe model can memorize the EEG examples."
        "\nThe architecture/backpropagation path is working."
    )

elif best_acc >= 70.0:
    print(
        "\nPARTIAL:"
        "\nThe model learns, but optimization is weak."
    )

else:
    print(
        "\nFAIL:"
        "\nThe model cannot memorize a tiny training set."
        "\nDo NOT run the 118-subject LOSO yet."
    )
```
